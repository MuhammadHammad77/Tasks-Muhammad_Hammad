```python
from sklearn.preprocessing import RobustScaler

rob_scaler = RobustScaler()

df['scaled_amount'] = rob_scaler.fit_transform(df['Amount'].values.reshape(-1,1))
df['scaled_time'] = rob_scaler.fit_transform(df['Time'].values.reshape(-1,1))

df.drop(['Time','Amount'], axis=1, inplace=True)
```

```python
from sklearn.model_selection import StratifiedShuffleSplit

# 1. Define your Features (X) and your Target (y)
X = df.drop('Class', axis=1)
y = df['Class']

# 2. Setup the Split (80% for training, 20% for testing)
# 'stratified' ensures the 99%/0.1% ratio is kept in both sets
sss = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)

for train_index, test_index in sss.split(X, y):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

# 3. Check if it worked
print('Training Set Fraud Ratio:', y_train.value_counts(normalize=True)[1])
print('Test Set Fraud Ratio:', y_test.value_counts(normalize=True)[1])
```

```python
from imblearn.over_sampling import SMOTE

# 1. Create the SMOTE tool
sm = SMOTE(sampling_strategy='minority', random_state=42)

# 2. Fit it ONLY to the training data
# This creates the synthetic fraud cases
X_train_res, y_train_res = sm.fit_resample(X_train, y_train)

# 3. Check the new balance
print('New Training Set Shape:', X_train_res.shape)
print('New Fraud vs Legitimate ratio:\n', y_train_res.value_counts(normalize=True))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score,recall_score,f1_score,precision_score
model=LogisticRegression(solver='liblinear')
model.fit(X_train_res,y_train_res)
y_pred=model.predict(X_test)
```

```python
model.fit(X_train_res, y_train_res)
y_pred=model.predict(X_test)
print(classification_report(y_test,y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)

rf_model.fit(X_train_res, y_train_res)
y_pred_rf = rf_model.predict(X_test)

print(" Random Forest Report ")
print(classification_report(y_test, y_pred_rf))
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

input_dim = X_train_res.shape[1]

model_dl = Sequential([

    Dense(32, input_shape=(input_dim,), activation='relu'),
    Dropout(0.2),

    Dense(16, activation='relu'),
    Dropout(0.2),

    Dense(1, activation='sigmoid')
])
model_dl.compile(optimizer='adam',
                 loss='binary_crossentropy',
                 metrics=['accuracy'])

model_dl.fit(X_train_res, y_train_res, epochs=10, batch_size=64, validation_split=0.1)
```

```python
import numpy as np
from sklearn.metrics import classification_report

raw_predictions = model_dl.predict(X_test)

y_pred_dl = (raw_predictions >= 0.5).astype(int)

print("--- Deep Learning (ANN) Report ---")
print(classification_report(y_test, y_pred_dl))
```
